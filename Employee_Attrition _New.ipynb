```python
num_cols = ['MonthlyIncome', 'YearsSinceLastPromotion', 
            'YearsAtCompany', 'YearsWithCurrManager']

for col in num_cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df[col], color='steelblue')
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
  
df[col] = np.where(df[col] < lower, lower,
                       np.where(df[col] > upper, upper, df[col]))
```

```python
factors = ['Department', 'Gender', 'MaritalStatus', 
           'BusinessTravel', 'EducationField', 'JobRole']

for col in factors:
    plt.figure(figsize=(7,4))
    sns.countplot(x=col, hue='Attrition', data=df)
    plt.title(f'{col.title()} Based on Attrition')
    plt.xticks(rotation=45)
    plt.show()
```

```python
def tenure_category(years):
    if years <= 2:
        return 'Short-term'
    elif 3 <= years <= 5:
        return 'Mid-term'
    else:
        return 'Long-term'

df['TenureCategory'] = df['YearsAtCompany'].apply(tenure_category)


df['PromotionGapRatio'] = df['YearsSinceLastPromotion'] / (df['TotalWorkingYears'] + 1)
```

```python

df['PerformanceToSalary'] = df['PerformanceRating'] / (df['MonthlyIncome'] + 1)


df['ExperiencePerformance'] = df['TotalWorkingYears'] * df['PerformanceRating']
```

```python
df['EngagementScore'] = (
    df['JobSatisfaction'] +
    df['EnvironmentSatisfaction'] +
    df['RelationshipSatisfaction'] +
    df['JobInvolvement']
) / 4
```
